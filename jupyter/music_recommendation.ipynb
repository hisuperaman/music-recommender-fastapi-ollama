recommended songs based on listening history and liked songs`

```python
import nltk
from nltk.stem.porter import PorterStemmer

nltk.download('punkt_tab')
stemmer = PorterStemmer()

def tokenization(txt):
    tokens = nltk.word_tokenize(txt)
    stemming = [stemmer.stem(w) for w in tokens]
    return " ".join(stemming)
```

```python
df['text'] = df['text'].apply(tokenization)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# TF-IDF for lyrics
tfidf = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf.fit_transform(df['text'])
lyrics_similarity = cosine_similarity(tfidf_matrix)
```

```python
from sklearn.preprocessing import MinMaxScaler

# Normalize audio features
features = ['Tempo', 'Energy', 'Positiveness']
scaler = MinMaxScaler()
df_scaled = scaler.fit_transform(df[features])
audio_similarity = cosine_similarity(df_scaled)
```

```python
combined_similarity = 0.6 * lyrics_similarity + 0.4 * audio_similarity
```

```python
def recommend_songs(user_history, liked_songs, top_n=20):
    # Normalize song names (strip spaces and lower case)
    user_history = [song.strip().lower() for song in user_history]
    liked_songs = [song.strip().lower() for song in liked_songs]
    all_known = user_history + liked_songs

    # Clean the song names in the dataframe
    df['song_clean'] = df['song'].str.strip().str.lower()

    # Find indices for songs in user history and liked songs
    indices = []
    for song in all_known:
        if song in df['song_clean'].values:
            idx = df[df['song_clean'] == song].index[0]
            indices.append(idx)

    if not indices:
        print("No matching songs found in dataset.")
        return []

    # Create a user profile using combined similarity
    user_profile = combined_similarity[indices].mean(axis=0)
    scores = list(enumerate(user_profile))
    scores = sorted(scores, key=lambda x: x[1], reverse=True)

    # Collect the recommendations
    recommended = []
    for idx, score in scores:
        song = df.iloc[idx]['song']
        artist = df.iloc[idx]['Artist(s)']
        if df.iloc[idx]['song_clean'] not in all_known:
            recommended.append((song, artist))
        if len(recommended) == top_n:
            break

    return recommended
```

```python
user_history = ['Lightwork Freestyle 2']
liked_songs = ['Down By The Water']

recommendations = recommend_songs(user_history, liked_songs)
for song, artist in recommendations:
    print(f"{song} by {artist}")
```

```python
def recommend_songs_by_vibes(vibes, top_n=20):
    fav_genre = vibes["favorite_genre"].strip().lower().replace(" ", "")
    emotion = vibes["dominant_emotion"]
    tempo = vibes["avg_tempo"]
    energy = vibes["avg_energy"]
    valence = vibes["avg_positiveness"]

    df['genre_clean'] = df['Genre'].fillna('').str.lower().str.replace(" ", "")
    df['song_clean'] = df['song'].str.strip().str.lower()

    filtered_df = df[
        df['genre_clean'].str.contains(fav_genre) &
        (df['emotion'] == emotion) &
        df['Tempo'].between(tempo - 0.1, tempo + 0.1) &
        df['Energy'].between(energy - 15, energy + 15) &
        df['Positiveness'].between(valence - 15, valence + 15)
    ]

    recommendations = filtered_df[['song', 'Artist(s)']].head(top_n).values.tolist()
    return recommendations
```

```python
import os
from spotipy import Spotify
from spotipy.oauth2 import SpotifyClientCredentials


client_id = "5ee39c99a4bd43fd9eb5492a6a049df9"
client_secret = "ae4befc0cf4445c8b890b69949788332"

sp = Spotify(auth_manager=SpotifyClientCredentials(client_id=client_id, client_secret=client_secret))

async def get_track_details(song: str, artist: str):
    query = f"track:{song} artist:{artist}"
    result = sp.search(q=query, type='track', limit=1)

    items = result.get("tracks", {}).get("items", [])
    if not items:
        return None

    track = items[0]


    return {
        "track_name": track["name"],
        "artist": track["artists"][0]["name"],
        "album": track["album"]["name"],
        "release_date": track["album"]["release_date"],
        "spotify_url": track["external_urls"]["spotify"],
        "preview_url": track["preview_url"],
        "cover_image": track["album"]["images"][0]["url"] if track["album"]["images"] else None,
    }
```

```python
import asyncio

# Assuming df has columns: 'song' and 'Artist(s)'
async def enrich_dataframe(df):
    data = []

    for i, row in df.iterrows():
        details = await get_track_details(row['song'], row['Artist(s)'])
        if details:
            data.append(details)
        else:
            data.append({
                "track_name": None,
                "artist": None,
                "album": None,
                "release_date": None,
                "spotify_url": None,
                "preview_url": None,
                "cover_image": None,
            })

        if i%100==0:
            print("processed", i)

    # Create a DataFrame from the fetched details
    details_df = pd.DataFrame(data)

    # Merge the details with the original DataFrame
    df = pd.concat([df.reset_index(drop=True), details_df[["album", "cover_image", "release_date", "spotify_url"]]], axis=1)

    return df
```

```python
def collaborative_filtering(df_likes, top_n=10):
    user_item_matrix = df_likes.pivot_table(index='user_email', columns='track_name', aggfunc='size', fill_value=0)

    similarity = cosine_similarity(user_item_matrix.T)
    similarity_df = pd.DataFrame(similarity, index=user_item_matrix.columns, columns=user_item_matrix.columns)

    def recommend_for_user(user_email):
        user_ratings = user_item_matrix.loc[user_email]
        liked_songs = user_ratings[user_ratings > 0].index

        scores = similarity_df[liked_songs].sum(axis=1)
        scores = scores.drop(labels=liked_songs)
        recommended = scores.sort_values(ascending=False).head(top_n)

        return list(recommended.index)

    return recommend_for_user
```
